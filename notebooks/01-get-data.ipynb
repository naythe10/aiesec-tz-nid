```python
import requests
import calendar
from pathlib import Path
import json
import os
```

```python
#Access token saved in env variables
token = os.environ["AIESEC_ACCESS_TOKEN"] #check .bashrc and api
base_url = "https://analytics.api.aiesec.org/v2/applications/analyze.json"
office_id = 567
```

```python
monthly_data = {}

for year in range(2022, 2026):
    for month in range(1, 13):
        start_date= f"{year}-{month:02d}-01" #2022-01-01
        last_day = calendar.monthrange(year, month)[1]
        end_date = f"{year}-{month:02d}-{last_day}"

        params = {
            "access_token": token,
            "performance_v3[office_id]": office_id,
            "start_date": start_date,
            "end_date": end_date,
        }

        response = requests.get(
            url= base_url,
            params= params
        )
        response.raise_for_status()

        monthly_data[f"{year}-{month:02d}"] = response.json()

print(f"Data acquired for {len(monthly_data)} months")    
```

```python
output_path = Path("../data/monthly_data.json")

with output_path.open("w", encoding= "utf-8") as file:
    json.dump(monthly_data, file, indent= 2)
```
