```python
import requests
import os
```

```python
#Access token saved in env variables
token = os.environ["AIESEC_ACCESS_TOKEN"]
```

```python
# Confirmed 567 is MC Tanzania, 
# from https://docs.google.com/spreadsheets/d/1A8Epd__j7OKFO85A0JAJUZcoCuvKee3DHpn0cDafx6E/edit?gid=0#gid=0

params = {
    "access_token": token,
    "performance_v3[office_id]" : 567, 
    "start_date" : "2022-01-01",
    "end_date" : "2025-12-31",
}

response = requests.get(
    url="https://analytics.api.aiesec.org/v2/applications/analyze.json",
    params= params
    )
```

```python
response.raise_for_status()
data = response.json()
```

```python
import json
with open("data.json", "w", encoding= "utf-8") as f:
    json.dump(data, f, indent= 4)
```
