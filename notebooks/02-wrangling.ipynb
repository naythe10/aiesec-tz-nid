```python
import os
import re
import pandas as pd
import json
from pathlib import Path
```

```python
data_dir = Path("../data/monthly_data.json")
```

```python
with data_dir.open("r", encoding= "utf-8") as file:
    data = json.load(file)
```

```python
rows = []

for month, raw in data.items():
    response_data = raw["response"]

    row = {
        "month": month,
        "applied": response_data["applied_total"]["applicants"]["value"],
        "matched": response_data["matched_total"]["applicants"]["value"],
        "accepted": response_data["an_accepted_total"]["applicants"]["value"],
        "approved": response_data["approved_total"]["applicants"]["value"],
        "realized": response_data["realized_total"]["applicants"]["value"],
        "completed": response_data["completed_total"]["applicants"]["value"],
    }
    rows.append(row)
```

```python
monthly_data = pd.DataFrame(rows)
monthly_data["month"] = pd.to_datetime(monthly_data["month"])
monthly_data.sort_values("month")
monthly_data.head(15)
```

```python
output_path = Path("../data/monthly_data.csv")
monthly_data.to_csv(output_path, index= False)
```
