```python
import pandas as pd
from pathlib import Path
```

```python
data_dir = Path('../data/monthly_data.csv')
data = pd.read_csv(data_dir)
data.head()
```

```python
data['month'] = pd.to_datetime(data['month']).dt.to_period("M")
data.set_index('month', inplace= True)
```

```python
data.head()
```

```python
data.info()
```

```python
# Monthly activity performance over the years
import matplotlib.pyplot as plt
plt.figure(figsize= (12,5))
plt.xlabel("Date")
plt.ylabel("Count")
for column in data.columns:
    plt.plot(data.index.to_timestamp(), data[column], label = column)
plt.title("AIESEC in Tanzania Monthly Exchange Funnel (2022-2025)")
plt.grid(True)
plt.legend()
plt.tight_layout()
```
